```python
# -------------------------------------------------------------------------
# INSIGHT: Missing Values in `product_weight_kg`
# -------------------------------------------------------------------------
# EDA reveals that the missingness in `product_weight_kg` is not completely
# random (MNAR/MAR). The vast majority of missing records originate from
# two specific stores: `STORE-T5G` and `STORE-7WS`.
#
# Finding:
# - These locations consistently fail to log product weights across their inventory,
#   pointing to an operational/systemic data capture issue at the store level
#   rather than product-level discrepancies.
#
# Handling Strategy:
# - Since product weight is an intrinsic physical property of an item (independent
#   of where it is sold), we should NOT impute using a simple overall mean/median.
# - Strategy: Impute missing weights by mapping `product_code` to known weights
#   recorded in other stores. For any remaining rare products with zero records,
#   fall back to the median weight of their respective `product_category`.
# -------------------------------------------------------------------------
```

### 📝 Data Quality Note: Missing `product_weight_kg`
- **Observation:** Missing values in `product_weight_kg` are heavily concentrated in two specific stores (`STORE-T5G` and `STORE-7WS`).
- **Diagnosis:** The missingness is structural/systemic to these store outlets (likely a logging or instrumentation gap during inventory entry), rather than random data loss.
- **Handling Strategy:** Avoid naive global mean/median imputation across the entire dataset, as it might distort store-level characteristics. Use product-level lookup (e.g., mapping `product_weight_kg` from other stores via `product_code`) or group-level medians.

```python
# -------------------------------------------------------------------------
# INSIGHT & DECISION: Missing Values in `store_size`
# -------------------------------------------------------------------------
# Observation:
# - `store_size` has 1,534 missing entries and shares strong domain overlap
#   with `store_format` and `store_location_tier`.
# - Missingness is concentrated systematically across specific store codes
#   rather than occurring at random row-level intervals.
#
# Evaluation (To Drop vs. To Keep):
# - While collinear with `store_format`, dropping `store_size` discards
#   granular scale signals that tree-based algorithms (LightGBM/XGBoost/CatBoost)
#   use effectively for split decisions and interaction features.
# - Unlike linear models, gradient-boosted trees are robust to multicollinearity
#   and can derive utility from structural missingness.
#
# Handling Plan:
# 1. Inspect the cross-tabulation of `(store_format, store_location_tier)` against `store_size`.
# 2. If a specific format-tier combination unambiguously determines size (e.g., Grocery Stores
#    consistently mapping to 'Small'), impute using that group mode.
# 3. For any ambiguous cases, retain the signal by encoding the nulls explicitly
#    as a distinct category: `'Missing'`.
# -------------------------------------------------------------------------
```

ways to fill missing values

```python
# 1. IMPUTE `product_weight_kg`
# Step 1a: Learn the median weight per product_code strictly from train_df
item_weight_map = train_df.dropna(subset=['product_weight_kg']).groupby('product_code')['product_weight_kg'].median().to_dict()

# Step 1b: Learn the category-level fallback median from train_df
cat_weight_map = train_df.dropna(subset=['product_weight_kg']).groupby('product_category')['product_weight_kg'].median().to_dict()

# Step 1c: Global fallback median (just in case)
global_weight_median = train_df['product_weight_kg'].median()

def impute_weight(df):
    # 1. Fill using product_code match
    df['product_weight_kg'] = df['product_weight_kg'].fillna(df['product_code'].map(item_weight_map))
    # 2. Fallback to product_category median
    df['product_weight_kg'] = df['product_weight_kg'].fillna(df['product_category'].map(cat_weight_map))
    # 3. Final safety net: global median
    df['product_weight_kg'] = df['product_weight_kg'].fillna(global_weight_median)
    return df

train_df = impute_weight(train_df)
val_df = impute_weight(val_df)

# 2. IMPUTE `store_size`
# Step 2a: Find the mode (most common size) per (store_format, store_location_tier)
def get_mode(s):
    mode_vals = s.mode()
    return mode_vals.iloc[0] if not mode_vals.empty else None

store_size_map = (
    train_df.dropna(subset=['store_size'])
    .groupby(['store_format', 'store_location_tier'])['store_size']
    .agg(get_mode)
    .to_dict()
)

def impute_store_size(df):
    # Create key tuple for fast lookup
    group_keys = list(zip(df['store_format'], df['store_location_tier']))
    mapped_sizes = pd.Series(group_keys, index=df.index).map(store_size_map)

    # 1. Fill using group mode mapping
    df['store_size'] = df['store_size'].fillna(mapped_sizes)
    # 2. Fill any remaining ambiguous values with explicit 'Missing' category
    df['store_size'] = df['store_size'].fillna('Missing')
    return df

train_df = impute_store_size(train_df)
val_df = impute_store_size(val_df)

# 3. VERIFICATION
print("Remaining nulls in train_df:")
print(train_df[['product_weight_kg', 'store_size']].isna().sum())

print("\nRemaining nulls in val_df:")
print(val_df[['product_weight_kg', 'store_size']].isna().sum())
```

```python
# STEP 1: ROW-LEVEL & STRING TRANSFORMATIONS (No Leakage Risk)
def add_row_features(df):
    # 1. Product broad family (e.g., FD -> Food, DR -> Drinks, NC -> Non-Consumable)
    df['broad_category'] = df['product_code'].str[:2]

    # 2. Value density: Price per unit weight
    df['price_per_kg'] = df['product_price'] / df['product_weight_kg']

    # 3. Store architecture interaction term
    df['store_format_size'] = df['store_format'].astype(str) + "_" + df['store_size'].astype(str)
    df['store_format_tier'] = df['store_format'].astype(str) + "_" + df['store_location_tier'].astype(str)

    return df

train_df = add_row_features(train_df)
val_df = add_row_features(val_df)

# STEP 2: GROUP AGGREGATION FEATURES (Compute on train_df, map to both)

# A. Category-level average price and visibility
cat_price_mean = train_df.groupby('product_category')['product_price'].mean().to_dict()
cat_visibility_mean = train_df.groupby('product_category')['shelf_visibility'].mean().to_dict()

# B. Store-level inventory variety (distinct products stocked)
store_variety = train_df.groupby('store_code')['product_code'].nunique().to_dict()

def add_aggregation_features(df):
    # Relative price within product category
    mapped_cat_price = df['product_category'].map(cat_price_mean)
    df['price_diff_from_cat_mean'] = df['product_price'] - mapped_cat_price
    df['price_ratio_to_cat_mean'] = df['product_price'] / mapped_cat_price

    # Relative shelf visibility within product category
    mapped_cat_vis = df['product_category'].map(cat_visibility_mean)
    df['visibility_ratio_to_cat'] = df['shelf_visibility'] / (mapped_cat_vis + 1e-6)

    # Store footprint proxy
    df['store_product_variety'] = df['store_code'].map(store_variety).fillna(0)

    return df

train_df = add_aggregation_features(train_df)
val_df = add_aggregation_features(val_df)

# STEP 3: SANITY CHECK
new_cols = [
    'broad_category', 'price_per_kg', 'store_format_size',
    'store_format_tier', 'price_diff_from_cat_mean',
    'visibility_ratio_to_cat', 'store_product_variety'
]

print("Engineered features created:")
print(train_df[new_cols].head())
```
